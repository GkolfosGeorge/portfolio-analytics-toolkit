```python
# ══ F. ENTRY STRATEGY / DCA ═══════════════════════════════════════════
RUN_THIS_CELL = RUN_DCA_BASIC or RUN_DCA_SMART
# ══════════════════════════════════════════════════════════════════════
if not RUN_THIS_CELL:
    print("Skipped.")
else:
    monthly_dates = dca.get_monthly_dates(prices.index)
    lump_vals     = (prices[TICKERS] *
                     (INITIAL_CAPITAL * weights / prices.iloc[0].values)).sum(axis=1)

    # F1 — Lump Sum vs Simple DCA (Tier 1)
    if RUN_DCA_BASIC:
        dca_vals, dca_inv = dca.simple_dca(
            prices[TICKERS], weights, monthly_dates, MONTHLY_DCA)
        print("── Lump Sum vs Simple DCA ───────────────────────────────")
        ch.plot_lump_vs_dca(lump_vals, dca_vals,
                             save_fig=SAVE_CHARTS, output_dir=OUTPUT_DIR)

    # F2/F3 — Smart DCA (Tier 2)
    if RUN_DCA_SMART:
        dd_vals, dd_inv = dca.drawdown_dca(
            prices[TICKERS], weights, monthly_dates, MONTHLY_DCA,
            window=DD_WINDOW, multiplier=DD_MULTIPLIER)
        ma_vals, ma_inv = dca.ma200_dca(
            prices[TICKERS], weights, monthly_dates, MONTHLY_DCA,
            ma_window=MA_WINDOW, tiers=MA_TIERS)
        print("── Smart DCA Strategy Comparison ────────────────────────")
        print(dca.dca_summary_metrics({
            "Simple DCA":   (dca_vals, dca_inv) if RUN_DCA_BASIC
                            else (dca.simple_dca(prices[TICKERS], weights,
                                  monthly_dates, MONTHLY_DCA)),
            "Drawdown DCA": (dd_vals,  dd_inv),
            "MA200 DCA":    (ma_vals,  ma_inv),
        }).to_string())
```

```python
# ══ G. MONTE CARLO & OPTIMIZATION ════════════════════════════════════
RUN_THIS_CELL = RUN_MONTE_CARLO
# ══════════════════════════════════════════════════════════════════════
if not RUN_THIS_CELL:
    print("Skipped — set RUN_MONTE_CARLO = True in CELL 0 to run.")
else:
    ef  = mc.efficient_frontier(
              prices, TICKERS, weights,
              num_portfolios=10_000, risk_free_rate=RISK_FREE_RATE)
    rw  = mc.multi_regime_optimization(prices, TICKERS, [
              {"label": "Full Period",
               "start": START_DATE, "end": str(prices.index[-1].date())},
              {"label": "Post-Covid",
               "start": "2020-06-01", "end": "2021-12-31"},
              {"label": "Rate Hike 2022",
               "start": "2022-01-01", "end": "2022-12-31"},
          ], num_portfolios=5_000)

    # ── Walk-Forward Test ───────────────────────────────────────────────
    total_days = len(prices)
    split_date = prices.index[int(total_days * 0.60)].strftime("%Y-%m-%d")
    wf = mc.walk_forward_test(
              prices, TICKERS,
              train_start    = START_DATE,
              train_end      = split_date,
              test_start     = split_date,
              test_end       = str(prices.index[-1].date()),
              num_portfolios = 5_000)
    wf_port  = (1 + wf["test_portfolio_ret"]).cumprod()
    wf_bench = (1 + wf["test_benchmark_ret"]).cumprod()
    wf_cagr  = float(wf_port.iloc[-1]  ** (252 / len(wf_port))  - 1)
    bm_cagr  = float(wf_bench.iloc[-1] ** (252 / len(wf_bench)) - 1)

    sim = mc.simulate_future_paths(
              portfolio_returns,
              horizon_days    = GOAL_YEARS * 252,
              num_simulations = 1_000,
              initial_value   = INITIAL_CAPITAL,
              percentiles     = (5, 25, 50, 75, 95))

    # ── Numbers ────────────────────────────────────────────────────────
    print("── Weight Comparison — Current vs Optimal ───────────────")
    print(mc.weight_comparison_table(TICKERS, weights, ef).to_string())
    print("\n── Regime-Optimal Weights (%) ───────────────────────────")
    print((rw * 100).round(1).to_string())
    print("\n── Walk-Forward Test (out-of-sample validation) ─────────")
    print(f"  Train : {START_DATE} → {split_date}")
    print(f"  Test  : {split_date} → {prices.index[-1].date()}")
    print(f"  Blind portfolio CAGR : {wf_cagr*100:.2f}%")
    print(f"  Benchmark CAGR       : {bm_cagr*100:.2f}%")
    print(f"  Alpha (out-of-sample): {(wf_cagr - bm_cagr)*100:+.2f}%")
    print(f"\n── Forward {GOAL_YEARS}-yr Simulation (1,000 paths) ─────────────")
    print(f"  p5     : €{sim['percentiles'][5][-1]:,.0f}")
    print(f"  Median : €{sim['percentiles'][50][-1]:,.0f}")
    print(f"  p95    : €{sim['percentiles'][95][-1]:,.0f}")

    # ── Charts ─────────────────────────────────────────────────────────
    ch.plot_efficient_frontier(ef,
                                save_fig=SAVE_CHARTS,
                                output_dir=OUTPUT_DIR)
    ch.plot_monte_carlo_paths(
        sim,
        initial_value = INITIAL_CAPITAL,
        goal_target   = GOAL_TARGET,
        horizon_years = GOAL_YEARS,
        save_fig      = SAVE_CHARTS,
        output_dir    = OUTPUT_DIR)
    ch.plot_monte_carlo_fan(
        sim,
        initial_value = INITIAL_CAPITAL,
        goal_target   = GOAL_TARGET,
        horizon_years = GOAL_YEARS,
        save_fig      = SAVE_CHARTS,
        output_dir    = OUTPUT_DIR)
```
